```python
import matplotlib.pyplot as plt

season_labels = ['Spring', 'Summer', 'Autumn', 'Winter']
season_means = df.groupby('season')['cnt'].mean()

plt.bar(season_labels, season_means, color='steelblue')
plt.xlabel('Season')
plt.ylabel('Average hourly bike hires')
plt.title('Average Bike-Share Ridership by Season')
plt.savefig('season_cnt.png', dpi=150, bbox_inches='tight')
plt.show()
```
